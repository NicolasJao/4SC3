We can use **.groupby** method to count the number of total incidents in each neighborhood and save the result in a dataframe called **df_Neighbourhood_Bicycle_Thefts** :


```python
df_Neighbourhood_Bicycle_Thefts = df.groupby(['NeighbourhoodName']).size().reset_index(name = 'Number of Bycicle Thefts')
print(df_Neighbourhood_Bicycle_Thefts)
```

**.sort_values** method is used to sort the values of a dataframe based on a column (ascending or decending):

```python
df_Neighbourhood_Bicycle_Thefts_sorted = df_Neighbourhood_Bicycle_Thefts.sort_values (by = 'Number of Bycicle Thefts', ascending = False)
print(df_Neighbourhood_Bicycle_Thefts_sorted)
```

**.head(10)** selects the top 10 rows on a dataframe:

```python
df_Neighbourhood_Bicycle_Thefts_sorted.head(10)
```

**.to_list** generates a list of Neighbourhood Names:


```python
unsafe_neighbourhoods = df_Neighbourhood_Bicycle_Thefts_sorted.head(10)['NeighbourhoodName'].to_list()
print('Ten top Unsafe naighbourhoods for bikes are ', end='')
for i in unsafe_neighbourhoods:
  print(i, ', ', end='')
```
